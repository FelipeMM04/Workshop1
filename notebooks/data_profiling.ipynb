```python
import pandas as pd
import numpy as np

# Cargar el dataset especificando el separador ';'
file_path = '../data/raw/candidates.csv'
df = pd.read_csv(file_path, sep=';')

# Mostrar las verdaderas dimensiones del dataset
print(f"Dimensiones correctas del dataset: {df.shape[0]} filas y {df.shape[1]} columnas\n")

# Mostrar las primeras 5 filas para verificar la estructura
df.head()
```

```python
# 1. Información de tipos de datos y valores no nulos
print("=== INFORMACIÓN GENERAL Y TIPOS DE DATOS ===")
df.info()

print("\n" + "="*50 + "\n")

# 2. Conteo de valores nulos por columna
print("=== VALORES NULOS POR COLUMNA ===")
print(df.isnull().sum())

print("\n" + "="*50 + "\n")

# 3. Conteo de registros duplicados exactos
print("=== REGISTROS DUPLICADOS ===")
print(f"Filas duplicadas encontradas: {df.duplicated().sum()}")
```

```python
# Estadísticas de YOE, Code Challenge Score y Technical Interview Score
print("=== ESTADÍSTICAS DESCRIPTIVAS ===")
df[['YOE', 'Code Challenge Score', 'Technical Interview Score']].describe()
```

```python
# 1. Rangos de los puntajes de evaluación y experiencia
print("=== RANGOS NUMÉRICOS (MIN Y MAX) ===")
display(df[['YOE', 'Code Challenge Score', 'Technical Interview Score']].describe())

print("\n" + "="*50 + "\n")

# 2. Conteo de Candidatos Contratados vs No Contratados según Regla de Negocio
# HIRED = Code Challenge Score >= 7 AND Technical Interview Score >= 7
hired_count = df[(df['Code Challenge Score'] >= 7) & (df['Technical Interview Score'] >= 7)].shape[0]
not_hired_count = len(df) - hired_count

print("=== EVALUACIÓN REGLA DE NEGOCIO (HIRED / NOT HIRED) ===")
print(f"Candidatos contratados (HIRED): {hired_count} ({hired_count/len(df)*100:.2f}%)")
print(f"Candidatos no contratados (NOT HIRED): {not_hired_count} ({not_hired_count/len(df)*100:.2f}%)")

print("\n" + "="*50 + "\n")

# 3. Valores únicos por variables categóricas principales
print("=== VALORES ÚNICOS CATEGÓRICOS ===")
print(f"Cantidad de Países distintos: {df['Country'].nunique()}")
print(f"Niveles de Seniority: {df['Seniority'].unique().tolist()}")
print(f"Cantidad de Tecnologías distintas: {df['Technology'].nunique()}")
```

```python
# Exportar métricas generales a un informe rápido
profiling_summary = pd.DataFrame({
    'Metric': [
        'Total Records', 
        'Total Columns', 
        'Duplicate Rows', 
        'Hired Candidates', 
        'Hired Ratio (%)',
        'Unique Countries',
        'Unique Technologies'
    ],
    'Value': [
        len(df), 
        df.shape[1], 
        df.duplicated().sum(), 
        hired_count, 
        round(hired_count / len(df) * 100, 2),
        df['Country'].nunique(),
        df['Technology'].nunique()
    ]
})

print("=== RESUMEN EJECUTIVO DE PROFILING ===")
display(profiling_summary)
```
